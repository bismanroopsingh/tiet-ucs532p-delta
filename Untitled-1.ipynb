```python


import os, warnings
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
warnings.filterwarnings('ignore')

from skimage.feature import local_binary_pattern, graycomatrix, graycoprops
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

DATA_DIR = r"D:\fabric\Fabric Defects Dataset\Fabric Defect Dataset"

CLASSES = {
    'defect free': 0,
    'hole':        1,
    'horizontal':  2,
    'lines':       3,
    'stain':       4,
    'Vertical':    5
}
CLASS_NAMES   = list(CLASSES.keys())
LABEL_TO_NAME = {v: k for k, v in CLASSES.items()}
IMAGE_SIZE    = (128, 128)
LBP_RADIUS    = 1
LBP_POINTS    = 8
RANDOM_SEED   = 42
print('Config ready!')
def extract_all_features(image_path):
    """
    Extracts LBP + GLCM features AND returns a dict of
    human-interpretable named values for explanation.
    """
    img  = cv2.imread(image_path)
    img  = cv2.resize(img, IMAGE_SIZE)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # ── LBP ─────────────────────────────────────────────
    lbp      = local_binary_pattern(gray, LBP_POINTS, LBP_RADIUS, method='uniform')
    n_bins   = LBP_POINTS + 2
    lbp_hist, _ = np.histogram(lbp.ravel(), bins=n_bins,
                               range=(0, n_bins), density=True)

    # ── GLCM at multiple distances/angles ────────────────
    glcm = graycomatrix(gray,
                        distances=[1, 2],
                        angles=[0, np.pi/4, np.pi/2, 3*np.pi/4],
                        levels=256, symmetric=True, normed=True)

    contrast      = graycoprops(glcm, 'contrast').mean()
    dissimilarity = graycoprops(glcm, 'dissimilarity').mean()
    homogeneity   = graycoprops(glcm, 'homogeneity').mean()
    energy        = graycoprops(glcm, 'energy').mean()
    correlation   = graycoprops(glcm, 'correlation').mean()

    # ── Edge density (Canny) ──────────────────────────────
    edges        = cv2.Canny(gray, 50, 150)
    edge_density = edges.sum() / (255.0 * gray.size)

    # ── LBP-derived texture measures ─────────────────────
    lbp_uniformity  = lbp_hist[-1] + lbp_hist[-2]   # uniform + near-uniform patterns
    lbp_complexity  = 1.0 - lbp_uniformity            # non-uniform = more complex texture
    lbp_entropy     = -np.sum(lbp_hist * np.log2(lbp_hist + 1e-10))  # randomness

    # ── Feature vector for classifier ─────────────────────
    glcm_vec   = []
    for prop in ['contrast','dissimilarity','homogeneity','energy','correlation']:
        glcm_vec.extend(graycoprops(glcm, prop).ravel())
    feature_vector = np.concatenate([lbp_hist, glcm_vec])

    # ── Named interpretable dict ───────────────────────────
    named = {
        'contrast':      contrast,
        'dissimilarity': dissimilarity,
        'homogeneity':   homogeneity,
        'energy':        energy,
        'correlation':   correlation,
        'edge_density':  edge_density,
        'lbp_uniformity': lbp_uniformity,
        'lbp_complexity': lbp_complexity,
        'lbp_entropy':    lbp_entropy,
    }
    return feature_vector, named, gray, lbp

print(' Feature extractor ready!')
def build_dataset():
    X, y, named_list, paths = [], [], [], []
    valid_ext = ('.png', '.jpg', '.jpeg', '.bmp', '.tif')
    for class_name, label in CLASSES.items():
        folder = os.path.join(DATA_DIR, class_name)
        if not os.path.exists(folder):
            print(f'  Not found: {folder}'); continue
        files = [f for f in os.listdir(folder) if f.lower().endswith(valid_ext)]
        print(f'   {class_name:15s}: {len(files)} images')
        for fname in files:
            path = os.path.join(folder, fname)
            try:
                fvec, named, _, _ = extract_all_features(path)
                X.append(fvec)
                y.append(label)
                named_list.append(named)
                paths.append(path)
            except: pass
    print(f'\n Loaded {len(X)} samples')
    return np.array(X), np.array(y), named_list, paths

print(' Loading dataset...')
X, y, named_list, image_paths = build_dataset()

# Split
(
    X_train, X_test,
    y_train, y_test,
    named_train, named_test,
    paths_train, paths_test
) = train_test_split(
    X, y, named_list, image_paths,
    test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

# Scale
scaler   = StandardScaler()
Xs_train = scaler.fit_transform(X_train)
Xs_test  = scaler.transform(X_test)

# Train
print('\n Training models...')
svm = SVC(kernel='rbf', C=10, gamma='scale',
          class_weight='balanced', probability=True, random_state=RANDOM_SEED)
svm.fit(Xs_train, y_train)

rf = RandomForestClassifier(n_estimators=200, class_weight='balanced',
                            random_state=RANDOM_SEED, n_jobs=-1)
rf.fit(Xs_train, y_train)

print(f'SVM Accuracy : {accuracy_score(y_test, svm.predict(Xs_test))*100:.2f}%')
print(f'RF  Accuracy : {accuracy_score(y_test, rf.predict(Xs_test))*100:.2f}%')
# ============================================================
#  DIAGNOSIS + FIX for "always predicts defect free" problem
#  Run this AFTER your original training block
# ============================================================

import numpy as np
from collections import Counter
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 1 — DIAGNOSE: See what's actually in your dataset
# ─────────────────────────────────────────────────────────────────────────────

print("=" * 55)
print("  DATASET DIAGNOSIS")
print("=" * 55)

# Count per class
train_counts = Counter(y_train)
test_counts  = Counter(y_test)

print("\n  Class distribution in TRAINING set:")
for label, count in sorted(train_counts.items()):
    bar = "█" * (count // 2)
    print(f"    {LABEL_TO_NAME[label]:15s}: {count:4d}  {bar}")

print("\n  Class distribution in TEST set:")
for label, count in sorted(test_counts.items()):
    bar = "█" * (count // 2)
    print(f"    {LABEL_TO_NAME[label]:15s}: {count:4d}  {bar}")

print("\n  Predictions on test set (RF):")
preds = rf.predict(Xs_test)
pred_counts = Counter(preds)
for label, count in sorted(pred_counts.items()):
    print(f"    {LABEL_TO_NAME[label]:15s}: predicted {count} times")

print("\n  Full classification report (RF):")
print(classification_report(y_test, preds,
      target_names=CLASS_NAMES, zero_division=0))

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 2 — CONFUSION MATRIX: See where the model is failing
# ─────────────────────────────────────────────────────────────────────────────

cm = confusion_matrix(y_test, preds)
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax)
ax.set_xlabel("Predicted")
ax.set_ylabel("True")
ax.set_title("Confusion Matrix — RF (before fix)")
plt.tight_layout()
plt.savefig("confusion_before_fix.png", dpi=150)
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 3 — FIX 1: Retrain with explicit class weights + tuned threshold
# ─────────────────────────────────────────────────────────────────────────────

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.utils.class_weight import compute_class_weight

# Compute class weights dynamically from actual distribution
classes_arr = np.array(sorted(train_counts.keys()))
weights_arr = compute_class_weight('balanced', classes=classes_arr, y=y_train)
class_weight_dict = dict(zip(classes_arr.tolist(), weights_arr.tolist()))

print("\n  Computed class weights:")
for label, w in class_weight_dict.items():
    print(f"    {LABEL_TO_NAME[label]:15s}: weight = {w:.3f}")

# Retrain RF with balanced weights and more trees
rf_fixed = RandomForestClassifier(
    n_estimators=300,
    class_weight=class_weight_dict,   # dynamic, not just 'balanced'
    max_depth=None,
    min_samples_leaf=1,
    random_state=RANDOM_SEED,
    n_jobs=-1
)
rf_fixed.fit(Xs_train, y_train)

preds_fixed = rf_fixed.predict(Xs_test)
print("\n  RF (fixed) classification report:")
print(classification_report(y_test, preds_fixed,
      target_names=CLASS_NAMES, zero_division=0))

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 4 — FIX 2: Threshold tuning (lower the bar for minority classes)
#  Instead of argmax(proba), use per-class thresholds
# ─────────────────────────────────────────────────────────────────────────────

def predict_with_thresholds(model, X, thresholds):
    """
    thresholds: dict {label: float}  — label is predicted if proba > threshold
    Lower threshold for minority classes so they are not drowned out.
    Falls back to argmax if no class clears its threshold.
    """
    probas = model.predict_proba(X)
    preds  = []
    for row in probas:
        # Score = proba / threshold  (relative confidence)
        scores = [row[i] / thresholds.get(i, 1.0)
                  for i in range(len(row))]
        preds.append(int(np.argmax(scores)))
    return np.array(preds)

# Minority classes get a lower threshold so they compete fairly
# Adjust these values if some classes are still missed
THRESHOLDS = {
    CLASSES['defect free']: 0.50,   # majority — keep normal threshold
    CLASSES['hole']:        0.30,
    CLASSES['horizontal']:  0.30,
    CLASSES['lines']:       0.30,
    CLASSES['stain']:       0.30,
    CLASSES['Vertical']:    0.30,
}

preds_thresh = predict_with_thresholds(rf_fixed, Xs_test, THRESHOLDS)
print("\n  RF (fixed + thresholds) classification report:")
print(classification_report(y_test, preds_thresh,
      target_names=CLASS_NAMES, zero_division=0))

# Confusion matrix after fix
cm2 = confusion_matrix(y_test, preds_thresh)
fig2, ax2 = plt.subplots(figsize=(8, 6))
sns.heatmap(cm2, annot=True, fmt='d', cmap='Greens',
            xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, ax=ax2)
ax2.set_xlabel("Predicted")
ax2.set_ylabel("True")
ax2.set_title("Confusion Matrix — RF (after fix)")
plt.tight_layout()
plt.savefig("confusion_after_fix.png", dpi=150)
plt.show()

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 5 — REPLACE the model used in XAI with the fixed one
# ─────────────────────────────────────────────────────────────────────────────

# From now on use rf_fixed + predict_with_thresholds
# Update the explain function to use thresholded prediction

import shap

explainer_fixed = shap.TreeExplainer(rf_fixed)

def explain_image(image_path, top_n=6):
    """
    Full pipeline: extract → threshold-predict → NL explanation + report card.
    Uses the fixed model.
    """
    fvec, named, gray, lbp = extract_all_features(image_path)
    fvec_scaled = scaler.transform(fvec.reshape(1, -1))

    # Threshold-based prediction
    probas     = rf_fixed.predict_proba(fvec_scaled)[0]
    scores     = [probas[i] / THRESHOLDS.get(i, 1.0) for i in range(len(probas))]
    pred_label = int(np.argmax(scores))
    pred_name  = LABEL_TO_NAME[pred_label]
    confidence = probas[pred_label] * 100

    # SHAP
    sv_all = explainer_fixed.shap_values(fvec_scaled)
    sv     = sv_all[0, :, pred_label]
    top_idx  = np.argsort(np.abs(sv))[::-1][:top_n]
    top_sv   = sv[top_idx]
    top_feat = [FEATURE_NAMES[i] for i in top_idx]
    top_val  = [fvec[i] for i in top_idx]

    # ── Console report ────────────────────────────────────────────────────────
    sep = "─" * 58
    print(f"\n{sep}")
    print(f"  FABRIC DEFECT ANALYSIS")
    print(f"  File       : {os.path.basename(image_path)}")
    print(sep)
    print(f"\n  VERDICT   :  {pred_name.upper()}")
    print(f"  CONFIDENCE:  {confidence:.1f}%")
    print(f"\n  BECAUSE the fabric had these features:\n")

    FEATURE_DESCRIPTIONS = {
        'contrast':      ("Contrast",       "sharp intensity jumps → likely structural damage (tear/hole)",
                                            "smooth transitions → intact weave"),
        'dissimilarity': ("Dissimilarity",  "high local variation → frayed or damaged region",
                                            "consistent local texture → intact weave"),
        'homogeneity':   ("Homogeneity",    "very uniform → normal weave",
                                            "irregular texture → weave disrupted by defect"),
        'energy':        ("Energy",         "highly ordered → clean fabric",
                                            "disordered structure → weave disrupted"),
        'correlation':   ("Correlation",    "strong directional regularity → aligned threads",
                                            "weak regularity → threads broken or displaced"),
        'LBP_bin':       ("Texture Pattern","this micro-pattern is dominant in defective regions",
                                            "this micro-pattern is dominant in clean fabric"),
    }

    def _group(name):
        for key in FEATURE_DESCRIPTIONS:
            if name.startswith(key): return key
        return None

    for fname, sv_val, rv in zip(top_feat, top_sv, top_val):
        group = _group(fname)
        if group is None: continue
        label, high_m, low_m = FEATURE_DESCRIPTIONS[group]
        direction = "HIGH" if rv > 0.5 else "LOW"
        meaning   = high_m if rv > 0.5 else low_m
        impact    = "↑ SUPPORTS this class" if sv_val > 0 else "↓ argues against"
        print(f"  • {label:18s} [{direction}  val={rv:.3f}]")
        print(f"    → {meaning}")
        print(f"    → SHAP: {impact}  (shap={sv_val:+.4f})\n")

    print(f"  Named measurements:")
    print(f"    Contrast    = {named['contrast']:.2f}")
    print(f"    Homogeneity = {named['homogeneity']:.3f}")
    print(f"    Energy      = {named['energy']:.4f}")
    print(f"    LBP Entropy = {named['lbp_entropy']:.4f}")
    print(f"    Edge Density= {named['edge_density']:.4f}")
    print(f"\n  All class probabilities:")
    for i, p in enumerate(probas):
        bar    = "█" * int(p * 35)
        marker = " ← PREDICTED" if i == pred_label else ""
        print(f"    {LABEL_TO_NAME[i]:15s} {bar:<35s} {p*100:5.1f}%{marker}")
    print(f"\n{sep}\n")

    # ── Visual report (reuse render_report_card from previous file) ───────────
    result = {
        'pred_name': pred_name, 'confidence': confidence,
        'probas': probas, 'named': named, 'gray': gray,
        'lbp': lbp, 'top_features': list(zip(top_feat, top_sv, top_val)), 'sv': sv
    }
    return result

# ─────────────────────────────────────────────────────────────────────────────
#  STEP 6 — Test: find one image of EACH class and explain it
# ─────────────────────────────────────────────────────────────────────────────

print("\n\n=== Explaining one image from each defect class ===")
seen_classes = set()
for path, true_label in zip(paths_test, y_test):
    if true_label not in seen_classes:
        seen_classes.add(true_label)
        print(f"\n[True label: {LABEL_TO_NAME[true_label]}]")
        explain_image(path)
    if len(seen_classes) == len(CLASS_NAMES):
        break
```
